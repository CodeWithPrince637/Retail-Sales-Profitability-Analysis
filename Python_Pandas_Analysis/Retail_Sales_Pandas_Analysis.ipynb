```python
# Create a separate copy for analysis

df_clean = df.copy()

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)

print("Duplicate rows:", df_clean.duplicated().sum())
print("Missing values:", df_clean.isnull().sum().sum())
```

```python
# Create useful date-based columns

df_clean['Order Year'] = df_clean['Order Date'].dt.year

df_clean['Order Month'] = df_clean['Order Date'].dt.month_name()

df_clean['Shipping Days'] = (
    df_clean['Ship Date'] - df_clean['Order Date']
).dt.days

df_clean[['Order Date', 'Order Year', 'Order Month',
          'Ship Date', 'Shipping Days']].head()
```

```python
# Calculate Profit Margin

df_clean['Profit Margin'] = (
    df_clean['Profit'] / df_clean['Sales']
) * 100

# Check the result
df_clean[['Sales', 'Profit', 'Profit Margin']].head()
```

```python
# Calculate Shipping Cost Percentage

df_clean['Shipping Cost %'] = (
    df_clean['Shipping Cost'] / df_clean['Sales']
) * 100

# Check the result
df_clean[['Sales', 'Shipping Cost', 'Shipping Cost %']].head()
```

```python
# Final check after feature engineering

print("Dataset Shape:", df_clean.shape)

print("\nMissing Values:")
display(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])

print("\nDuplicate Rows:", df_clean.duplicated().sum())

print("\nNegative Shipping Days:",
      (df_clean['Shipping Days'] < 0).sum())

print("\nInvalid Profit Margin:",
      df_clean['Profit Margin'].isnull().sum())

print("\nInvalid Shipping Cost %:",
      df_clean['Shipping Cost %'].isnull().sum())
```

```python
# Overall Business Performance

total_sales = df_clean['Sales'].sum()
total_profit = df_clean['Profit'].sum()
total_orders = df_clean['Order ID'].nunique()
total_customers = df_clean['Customer ID'].nunique()
total_quantity = df_clean['Quantity'].sum()

overall_margin = (total_profit / total_sales) * 100

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Quantity:", total_quantity)
print("Overall Profit Margin:", round(overall_margin, 2), "%")
```

```python
import pandas as pd

df = pd.read_excel(path, sheet_name='Sheet1')

print("Dataset Shape:", df.shape)
print("Unique Row IDs:", df['Row ID'].nunique())
```

```python
print("Total Columns:", len(df.columns))

for i, col in enumerate(df.columns, 1):
    print(i, col)
```

```python

# Keep only the original 24 dataset columns
df = df.iloc[:, :24].copy()

# Convert original date columns
df['Order Date'] = pd.to_datetime(
    df['Order Date'], dayfirst=True
)

df['Ship Date'] = pd.to_datetime(
    df['Ship Date'], dayfirst=True
)

# Create analysis copy
df_clean = df.copy()

# Feature Engineering
df_clean['Order Year'] = df_clean['Order Date'].dt.year
df_clean['Order Month'] = df_clean['Order Date'].dt.month_name()

df_clean['Shipping Days'] = (
    df_clean['Ship Date'] - df_clean['Order Date']
).dt.days

df_clean['Profit Margin'] = (
    df_clean['Profit'] / df_clean['Sales']
) * 100

df_clean['Shipping Cost %'] = (
    df_clean['Shipping Cost'] / df_clean['Sales']
) * 100

# Final verification
print("Original Dataset:", df.shape)
print("Analysis Dataset:", df_clean.shape)
print("Unique Row IDs:", df_clean['Row ID'].nunique())
print("Duplicate Rows:", df_clean.duplicated().sum())
print("Missing Values:")
display(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```

```python

yearly_analysis = (
    df_clean.groupby('Order Year')
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum'),
        Total_Orders=('Order ID', 'nunique'),
        Total_Quantity=('Quantity', 'sum')
    )
)

yearly_analysis['Profit_Margin_%'] = (
    yearly_analysis['Total_Profit'] /
    yearly_analysis['Total_Sales'] * 100
)

yearly_analysis['Sales_Growth_%'] = (
    yearly_analysis['Total_Sales'].pct_change() * 100
)

display(yearly_analysis.round(2))
```

year wise sales and profit

```python

# Month-wise Sales and Profit Analysis

monthly_analysis = (
    df_clean.groupby(
        df_clean['Order Date'].dt.to_period('M')
    )
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum'),
        Total_Orders=('Order ID', 'nunique'),
        Total_Quantity=('Quantity', 'sum')
    )
)

monthly_analysis['Profit_Margin_%'] = (
    monthly_analysis['Total_Profit'] /
    monthly_analysis['Total_Sales'] * 100
)

monthly_analysis['Sales_Growth_%'] = (
    monthly_analysis['Total_Sales'].pct_change() * 100
)

display(monthly_analysis.round(2))


# Identify best and worst sales months

best_month = monthly_analysis['Total_Sales'].idxmax()
worst_month = monthly_analysis['Total_Sales'].idxmin()

print("Highest Sales Month:", best_month)
print("Highest Monthly Sales:", round(
    monthly_analysis.loc[best_month, 'Total_Sales'], 2
))

print("\nLowest Sales Month:", worst_month)
print("Lowest Monthly Sales:", round(
    monthly_analysis.loc[worst_month, 'Total_Sales'], 2
))
```

```python

# Category-wise Analysis

category_analysis = (
    df_clean.groupby('Category')
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum'),
        Total_Orders=('Order ID', 'nunique'),
        Total_Quantity=('Quantity', 'sum')
    )
    .sort_values('Total_Sales', ascending=False)
)

category_analysis['Profit_Margin_%'] = (
    category_analysis['Total_Profit'] /
    category_analysis['Total_Sales'] * 100
)

print("CATEGORY-WISE PERFORMANCE")
display(category_analysis.round(2))


# Sub-category-wise Analysis

subcategory_analysis = (
    df_clean.groupby('Sub-Category')
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum'),
        Total_Orders=('Order ID', 'nunique'),
        Total_Quantity=('Quantity', 'sum')
    )
    .sort_values('Total_Profit', ascending=False)
)

subcategory_analysis['Profit_Margin_%'] = (
    subcategory_analysis['Total_Profit'] /
    subcategory_analysis['Total_Sales'] * 100
)

print("\nSUB-CATEGORY-WISE PERFORMANCE")
display(subcategory_analysis.round(2))

print("\nTop 5 Profitable Sub-Categories")
display(subcategory_analysis.head(5).round(2))

print("\nBottom 5 Sub-Categories by Profit")
display(subcategory_analysis.tail(5).round(2))
```

category and subcategory wise analysis

```python

# Region-wise Business Analysis

region_analysis = (
    df_clean.groupby('Region')
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum'),
        Total_Orders=('Order ID', 'nunique'),
        Total_Quantity=('Quantity', 'sum')
    )
    .sort_values('Total_Sales', ascending=False)
)

region_analysis['Profit_Margin_%'] = (
    region_analysis['Total_Profit'] /
    region_analysis['Total_Sales'] * 100
)

print("REGION-WISE PERFORMANCE")
display(region_analysis.round(2))


# Market-wise Business Analysis

market_analysis = (
    df_clean.groupby('Market')
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum'),
        Total_Orders=('Order ID', 'nunique'),
        Total_Quantity=('Quantity', 'sum')
    )
    .sort_values('Total_Sales', ascending=False)
)

market_analysis['Profit_Margin_%'] = (
    market_analysis['Total_Profit'] /
    market_analysis['Total_Sales'] * 100
)

print("\nMARKET-WISE PERFORMANCE")
display(market_analysis.round(2))


# Identify highest and lowest profit regions

print("\nMost Profitable Region:",
      region_analysis['Total_Profit'].idxmax())

print("Lowest Profit Region:",
      region_analysis['Total_Profit'].idxmin())
```

region wise

```python

# Customer-wise Analysis

customer_analysis = (
    df_clean.groupby(['Customer ID', 'Customer Name'])
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum'),
        Total_Orders=('Order ID', 'nunique'),
        Total_Quantity=('Quantity', 'sum')
    )
    .sort_values('Total_Sales', ascending=False)
)

customer_analysis['Profit_Margin_%'] = (
    customer_analysis['Total_Profit'] /
    customer_analysis['Total_Sales'] * 100
)

print("TOP 10 CUSTOMERS BY SALES")
display(customer_analysis.head(10).round(2))


# Product-wise Analysis

product_analysis = (
    df_clean.groupby(['Product ID', 'Product Name'])
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum'),
        Total_Orders=('Order ID', 'nunique'),
        Total_Quantity=('Quantity', 'sum')
    )
    .sort_values('Total_Profit', ascending=False)
)

product_analysis['Profit_Margin_%'] = (
    product_analysis['Total_Profit'] /
    product_analysis['Total_Sales'] * 100
)

print("\nTOP 10 PRODUCTS BY PROFIT")
display(product_analysis.head(10).round(2))

print("\nBOTTOM 5 PRODUCTS BY PROFIT")
display(product_analysis.tail(5).round(2))
```

```python

# Discount-wise Analysis

discount_analysis = (
    df_clean.groupby('Discount')
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum'),
        Total_Orders=('Order ID', 'nunique'),
        Total_Quantity=('Quantity', 'sum')
    )
    .sort_index()
)

discount_analysis['Profit_Margin_%'] = (
    discount_analysis['Total_Profit'] /
    discount_analysis['Total_Sales'] * 100
)

print("DISCOUNT-WISE PERFORMANCE")
display(discount_analysis.round(2))


# Shipping Mode Analysis

shipping_analysis = (
    df_clean.groupby('Ship Mode')
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum'),
        Total_Orders=('Order ID', 'nunique'),
        Total_Shipping_Cost=('Shipping Cost', 'sum'),
        Average_Shipping_Days=('Shipping Days', 'mean')
    )
    .sort_values('Total_Sales', ascending=False)
)

shipping_analysis['Profit_Margin_%'] = (
    shipping_analysis['Total_Profit'] /
    shipping_analysis['Total_Sales'] * 100
)

print("\nSHIPPING MODE PERFORMANCE")
display(shipping_analysis.round(2))


# Overall Shipping Cost Impact

print("\nTOTAL SHIPPING COST:",
      round(df_clean['Shipping Cost'].sum(), 2))

print("AVERAGE SHIPPING DAYS:",
      round(df_clean['Shipping Days'].mean(), 2))

print("AVERAGE SHIPPING COST:",
      round(df_clean['Shipping Cost'].mean(), 2))
```

discound and shipping analysis

```python

# 1. Loss-making Sub-Categories

loss_subcategories = subcategory_analysis[
    subcategory_analysis['Total_Profit'] < 0
].sort_values('Total_Profit')

print("LOSS-MAKING SUB-CATEGORIES")
display(loss_subcategories.round(2))


# 2. Order-wise Profitability Analysis

order_analysis = (
    df_clean.groupby('Order ID')
    .agg(
        Order_Sales=('Sales', 'sum'),
        Order_Profit=('Profit', 'sum')
    )
)

loss_orders = order_analysis[
    order_analysis['Order_Profit'] < 0
]

print("\nORDER PERFORMANCE")
print("Total Orders:", len(order_analysis))
print("Loss-making Orders:", len(loss_orders))
print(
    "Loss-making Order Percentage:",
    round(len(loss_orders) / len(order_analysis) * 100, 2),
    "%"
)


# 3. Customer Segment Analysis

segment_analysis = (
    df_clean.groupby('Segment')
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum'),
        Total_Orders=('Order ID', 'nunique'),
        Total_Customers=('Customer ID', 'nunique')
    )
    .sort_values('Total_Profit', ascending=False)
)

segment_analysis['Profit_Margin_%'] = (
    segment_analysis['Total_Profit'] /
    segment_analysis['Total_Sales'] * 100
)

print("\nCUSTOMER SEGMENT PERFORMANCE")
display(segment_analysis.round(2))
```

```python

import pandas as pd
from google.colab import files

output_path = '/content/Retail_Sales_Pandas_Analysis.xlsx'

with pd.ExcelWriter(output_path, engine='openpyxl') as writer:

    # Cleaned dataset
    df_clean.to_excel(
        writer, sheet_name='Cleaned_Data', index=False
    )

    # Business analysis results
    yearly_analysis.to_excel(writer, sheet_name='Yearly_Analysis')
    monthly_analysis.to_excel(writer, sheet_name='Monthly_Analysis')
    category_analysis.to_excel(writer, sheet_name='Category_Analysis')
    subcategory_analysis.to_excel(writer, sheet_name='Subcategory_Analysis')
    region_analysis.to_excel(writer, sheet_name='Region_Analysis')
    market_analysis.to_excel(writer, sheet_name='Market_Analysis')
    discount_analysis.to_excel(writer, sheet_name='Discount_Analysis')
    shipping_analysis.to_excel(writer, sheet_name='Shipping_Analysis')
    segment_analysis.to_excel(writer, sheet_name='Segment_Analysis')

    # Top performers and business insights
    customer_analysis.head(10).to_excel(
        writer, sheet_name='Top_10_Customers'
    )
    product_analysis.head(10).to_excel(
        writer, sheet_name='Top_10_Products'
    )
    loss_subcategories.to_excel(
        writer, sheet_name='Loss_Subcategories'
    )
    order_analysis.to_excel(
        writer, sheet_name='Order_Analysis'
    )

print("Excel report created successfully!")

files.download(output_path)
```
